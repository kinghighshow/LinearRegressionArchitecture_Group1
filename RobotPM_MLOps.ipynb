- Separate major machine learning responsibilities into reusable Python modules.
- Store experiment parameters in a YAML configuration file rather than hard-coding them.
- Support reproducible data preprocessing and model training.
- Compare from-scratch and Scikit-learn model implementations using the same workflow.
- Track experiment parameters and evaluation metrics outside the notebooks.
- Make the project easier to test, maintain, and extend.

## Configuration-Driven Design

Experiment settings are stored in `configs/experiment_config.yaml`. Keeping these values in a configuration file separates experiment settings from the Python implementation and makes experiments easier to modify and reproduce.

For the housing linear regression workflow, the configuration specifies:

```yaml
modeling:
  feature: median_income
  target: median_house_value

training:
  test_size: 0.2
  random_state: 42
  learning_rate: 0.01
  iterations: 1000
```

The configuration also contains settings for the project's CSV, API, and database data sources.

This approach reduces hard-coded values in the pipeline. A user can change experiment parameters in the YAML configuration without rewriting the model implementation.

## Experiment Tracking

Model results are stored outside the notebook so that experiment results persist after execution.

The housing experiment records information such as:

- model implementation
- predictor
- learning rate
- number of iterations
- train/test split
- RMSE
- MAE
- R²

The current housing experiment produced the following results:

Saving experiment results to CSV provides a persistent record that can be reviewed and compared across model runs.

## Reproducibility Approach

The project uses several practices to make experiments easier to reproduce:

- Project dependencies are documented in `requirements.txt`.
- A dedicated Python virtual environment can be created for the project.
- Jupyter notebooks use the project virtual environment as their kernel.
- Experiment settings are centralized in `configs/experiment_config.yaml`.
- The housing train/test split uses a fixed `random_state` of `42`.
- Preprocessing fits the scaler on the training data and applies the fitted scaler to the test data.
- Both regression implementations use the same prepared training and test data.
- Evaluation results are saved to CSV for later comparison.
- Source code is separated into reusable modules rather than being implemented only inside notebooks.

Together, these practices help ensure that another team member can recreate the environment, rerun the workflow, and compare results using the same configuration.
